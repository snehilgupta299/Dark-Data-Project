```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("../excel/keywords.xlsx")
print(df.head())
```

```python
df.plot(kind="bar", x="word", y="count", legend=False)
plt.title("Most Common Issues in Support Tickets")
plt.xlabel("Keyword")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```
